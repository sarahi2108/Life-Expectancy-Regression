```python
"""
eval.py

Evaluate a saved model on feature_engineered_eval.csv.
"""

from pathlib import Path
import json

import joblib
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


PROC_DIR = Path("data/processed")
ARTIFACT_DIR = Path("artifacts")


def evaluate_model(
    model_path: Path = ARTIFACT_DIR / "model_xgb.pkl",
    eval_path: Path = PROC_DIR / "feature_engineered_eval.csv",
    target: str = "Life Expectancy",
    metrics_output: Path = ARTIFACT_DIR / "eval_metrics.json",
):
    eval_df = pd.read_csv(eval_path)
    X_eval = eval_df.drop(columns=[target])
    y_eval = eval_df[target]

    model = joblib.load(model_path)

    preds = model.predict(X_eval)

    rmse = mean_squared_error(y_eval, preds)
    mae = mean_absolute_error(y_eval, preds)
    r2 = r2_score(y_eval, preds)

    print("=== Eval metrics ===")
    print(f"RMSE: {rmse:.4f}")
    print(f"MAE:  {mae:.4f}")
    print(f"R²:   {r2:.4f}")

    metrics = {"rmse": rmse, "mae": mae, "r2": r2}
    with open(metrics_output, "w") as f:
        json.dump(metrics, f, indent=4)

    print(f"✅ Metrics saved to {metrics_output}")
    return metrics


if __name__ == "__main__":
    evaluate_model()
```
