```python
"""
preprocess.py

Mirror the cleaning logic of 01_EDA_cleaning:
- Load data/raw/train.csv, eval.csv, holdout.csv
- Drop identifier columns like FIPS and County
- (Optionally) apply any extra column-level cleaning you had
- Save cleaned splits to data/processed/cleaning_*.csv
"""

from pathlib import Path
from typing import Tuple

import pandas as pd


RAW_DIR = Path("data/raw")
PROC_DIR = Path("data/processed")
PROC_DIR.mkdir(parents=True, exist_ok=True)


def drop_fips_and_county(df: pd.DataFrame) -> pd.DataFrame:
    """
    From 01_EDA_cleaning:
    Drop FIPS and County columns (and any other IDs you had).
    """
    cols_to_drop = ["FIPS", "County"]
    existing = [c for c in cols_to_drop if c in df.columns]
    return df.drop(columns=existing)


def clean_splits(
    train_path: Path = RAW_DIR / "train.csv",
    eval_path: Path = RAW_DIR / "eval.csv",
    holdout_path: Path = RAW_DIR / "holdout.csv",
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Apply your EDA cleaning logic, then save under data/processed/cleaning_*.csv.
    """

    train_df = pd.read_csv(train_path)
    eval_df = pd.read_csv(eval_path)
    holdout_df = pd.read_csv(holdout_path)

    # 1. Drop FIPS / County (from notebook)
    train_df = drop_fips_and_county(train_df)
    eval_df = drop_fips_and_county(eval_df)
    holdout_df = drop_fips_and_county(holdout_df)

    # 2. Any *extra* cleaning logic you had (types, missing values, etc.)
    #    If your notebook had explicit operations (e.g. replace(), fillna(), etc.),
    #    you can add them here so all three splits get the same treatment.
    #
    # Example pattern:
    # cols_to_float = [ ... ]
    # for col in cols_to_float:
    #     train_df[col] = train_df[col].astype(float)
    #     eval_df[col] = eval_df[col].astype(float)
    #     holdout_df[col] = holdout_df[col].astype(float)

    # Save cleaned splits
    train_clean = PROC_DIR / "cleaning_train.csv"
    eval_clean = PROC_DIR / "cleaning_eval.csv"
    holdout_clean = PROC_DIR / "cleaning_holdout.csv"

    train_df.to_csv(train_clean, index=False)
    eval_df.to_csv(eval_clean, index=False)
    holdout_df.to_csv(holdout_clean, index=False)

    print("✅ Saved cleaned splits to data/processed/:")
    print("   cleaning_train:", train_df.shape, "->", train_clean)
    print("   cleaning_eval:", eval_df.shape, "->", eval_clean)
    print("   cleaning_holdout:", holdout_df.shape, "->", holdout_clean)

    return train_df, eval_df, holdout_df


if __name__ == "__main__":
    clean_splits()
```
